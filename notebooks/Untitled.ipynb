## Data loading pipeline

```python
import os


data_path = '../Data_Flow_JBHI'
```

```python
device_value = 'phone'
ground_truth = []
files = []

for v in os.listdir(data_path):
    v_dir = os.path.join(data_path, v)
    if not v_dir.endswith('.xlsx'):
        for sub_v in os.listdir(v_dir):
            sub_v_dir = os.path.join(v_dir, sub_v)
            if sub_v_dir.endswith('.csv'):
                ground_truth.append(sub_v_dir)
            elif device_value in sub_v_dir:
                files.append(sub_v_dir)
            else:
                continue        
```
